```python
# Load necessary libraries
import pandas as pd
import numpy as np
```

```python
# Read in data
world_indicators = pd.read_csv("rawData/World_Development_Indicator/Indicator_Data_1960-2023.csv")
country_medals = pd.read_csv("rawData/Sum_Win_Olympics_1896-2022/Olympic_Games_Medal_Tally.csv")
```

```python
# Drop unnecessary column
world_indicators = world_indicators.drop(columns=['Series Code'])

# Drop regions and keep countries only
world_indicators = world_indicators[:1519]

# Fill missing values
world_indicators = world_indicators.replace('..', np.nan)
```

```python
world_indicators.tail()
```

```python
# Get number of countries
len(world_indicators['Country Name'].unique().tolist())
```

```python
# Convert year columns into rows
world_indicators = world_indicators.melt(id_vars=['Country Name', 'Country Code', 'Series Name'], var_name='Year')

world_indicators['value'] = pd.to_numeric(world_indicators['value'], errors='coerce')

world_indicators.head(20)
```

```python
# Convert series name rows into columns
world_indicators = world_indicators.pivot_table(index=['Country Name', 'Country Code', 'Year'], columns='Series Name', values='value').reset_index().rename_axis(columns=None)

world_indicators.head(20)
```

```python
# Clean year column
world_indicators['Year'] = world_indicators['Year'].str.extract(r'(\d{4})').astype(int)

world_indicators.head()
```

```python
print(country_medals.shape)
print(world_indicators.shape)
```

```python
# Extract season column
season_pattern = r'(Summer|Winter|Equestrian|Intercalated)'
country_medals['season'] = country_medals['edition'].str.extract(season_pattern)

# Drop extra columns
country_medals = country_medals.drop(columns=['edition',
                                              'edition_id'])

country_medals.head()

country_medals.to_csv('country_medals.csv', index=False)
```

```python
# Convert to types
country_medals['country_noc'] = country_medals['country_noc'].astype(str)
world_indicators['Country Code'] = world_indicators['Country Code'].astype(str)

country_medals['country_noc'] = country_medals['country_noc'].str.strip()
world_indicators['Country Code'] = world_indicators['Country Code'].str.strip()

country_medals['year'] = country_medals['year'].astype(int)
world_indicators['Year'] = world_indicators['Year'].astype(int)
```

```python
# Get a list of unmatched country codes used for the same countries
country_medal_codes = country_medals['country_noc'].unique().tolist()

world_indicator_codes = world_indicators['Country Code'].unique().tolist()

unmatched_codes = list(set(country_medal_codes) - set(world_indicator_codes))

print(unmatched_codes)
print(len(unmatched_codes))
```

```python
# Replace only the matching values and keep all unmachted values unchanged
world_indicators['Country Code'] = world_indicators['Country Code'].replace(code_dict)
```

```python
# Get a list of unmatched country codes used for the same countries
country_medal_codes = country_medals['country_noc'].unique().tolist()

world_indicator_codes = world_indicators['Country Code'].unique().tolist()

unmatched_codes = list(set(country_medal_codes) - set(world_indicator_codes))

print(unmatched_codes)
print(len(unmatched_codes))
```

```python
# Merge datasets
final_medal_indicator_dataset = pd.merge(
    country_medals,
    world_indicators,
    left_on=['country_noc', 'year'],
    right_on=['Country Code', 'Year'],
    how='left'
)

print(final_medal_indicator_dataset.shape[0])
```

```python
# Get a list of unmatched country codes used for the same countries
country_medal_codes = country_medals['country_noc'].unique().tolist()

final_medal_indicator_data_codes = final_medal_indicator_dataset['Country Code'].unique().tolist()

unmatched_codes = list(set(country_medal_codes) - set(final_medal_indicator_data_codes))

print(unmatched_codes)
print(len(unmatched_codes))
```

```python
# Drop extra columns
final_medal_indicator_dataset = final_medal_indicator_dataset.drop(columns=['Year',
                                                                            'Country Code',
                                                                            'Country Name'])

# Rename columns
final_medal_indicator_dataset = final_medal_indicator_dataset.rename(columns={'year': 'Year',
                                                                              'season': 'Season',
                                                                              'country': 'Country',
                                                                              'country_noc': 'Country Code',
                                                                              'gold': 'Gold',
                                                                              'silver': 'Silver',
                                                                              'bronze': 'Bronze',
                                                                              'total': 'Total',
                                                                              'GDP per capita (current US$)': 'GDP Per Capita (current US$)',
                                                                              'Population, total': 'Total Population',
                                                                              'Life expectancy at birth, total (years)': 'Life Expectancy At Birth (years)',
                                                                              'Adjusted savings: education expenditure (current US$)': 'Education Expenditure (current US$)',
                                                                              'Current health expenditure per capita (current US$)': 'Health Expenditure Per Capita (current US$)',
                                                                              'Human capital index (HCI) (scale 0-1)': 'Human Capital Index (HCI) (scale 0-1)'})

# Reoder columns
final_medal_indicator_dataset = final_medal_indicator_dataset[['Year',
                                                               'Season',
                                                               'Country',
                                                               'Country Code',
                                                               'Gold',
                                                               'Silver',
                                                               'Bronze',
                                                               'Total',
                                                               'GDP (current US$)',
                                                               'GDP Per Capita (current US$)',
                                                               'Total Population',
                                                               'Life Expectancy At Birth (years)',
                                                               'Education Expenditure (current US$)',
                                                               'Health Expenditure Per Capita (current US$)',
                                                               'Human Capital Index (HCI) (scale 0-1)']]
```
