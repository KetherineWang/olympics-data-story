```python
import pandas as pd
import numpy as np
import re
```

```python
# read in data
athlete_bio = pd.read_csv("rawData/archive/Olympic_Athlete_Bio.csv")
athlete_event_results = pd.read_csv("rawData/archive/Olympic_Athlete_Event_Results.csv")
country_medal_count = pd.read_csv("rawData/archive/Olympic_Games_Medal_Tally.csv")
countries = pd.read_csv("rawData/archive/Olympics_Country.csv")
```

```python
# merge data sets, drop columns
complete_athlete_info = pd.merge(athlete_event_results, athlete_bio, on='athlete_id', how='left')
complete_athlete_info = complete_athlete_info.drop(columns=['country_noc_x', 
                                                            'edition_id',
                                                            'result_id',
                                                            'event',
                                                            'pos',
                                                            'isTeamSport',
                                                            'height',
                                                            'weight',
                                                            'description',
                                                            'special_notes',
                                                            'name'])
complete_athlete_info = complete_athlete_info.rename(columns={'country_noc_y': 'country_noc'})
```

```python
complete_athlete_info.head()
```

```python
# year column
year_pattern = r'(\d{4})'
complete_athlete_info['year'] = complete_athlete_info['edition'].str.extract(year_pattern)

# season column
season_pattern = r'(Summer|Winter)'
complete_athlete_info['season'] = complete_athlete_info['edition'].str.extract(season_pattern)
```

```python
# drop edition column
complete_athlete_info = complete_athlete_info.drop(columns='edition')
```
