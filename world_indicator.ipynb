```python
import pandas as pd
import numpy as np
```

```python
# load in data
world_indicators = pd.read_csv("rawData/World_Development_Indicator/Indicator_Data_1960-2023.csv")
country_medal_count = pd.read_csv("rawData/Sum_Win_Olympics_1896-2022/Olympic_Games_Medal_Tally.csv")
```

```python
# drop column
world_indicators = world_indicators.drop(columns=['Series Code', 'Country Name'])
# drop regions
world_indicators = world_indicators[:1519]
```

```python
# sub in NaN
world_indicators = world_indicators.replace('..', np.nan)
```

```python
# years into rows
world_indicators = world_indicators.melt(id_vars=['Country Code', 'Series Name'], var_name='Year')
world_indicators['value'] = pd.to_numeric(world_indicators['value'], errors='coerce')
world_indicators.head(20)
```

```python
# series name into columns
world_indicators = world_indicators.pivot_table(index=['Country Code', 'Year'], columns='Series Name', values='value')
world_indicators.head(10)
```

```python
# reset index
world_indicators = world_indicators.reset_index().rename_axis(columns=None)
world_indicators.head()
```

```python
# extract year
world_indicators['Year'] = world_indicators['Year'].str.extract(r'(\d{4})').astype(int)
world_indicators.head()
```

```python
print(country_medal_count.shape)
print(world_indicators.shape)
```

```python
# extract season
season_pattern = r'(Summer|Winter|Equestrian|Intercalated)'
country_medal_count['season'] = country_medal_count['edition'].str.extract(season_pattern)
country_medal_count = country_medal_count.drop(columns='edition')
country_medal_count.head()
```

```python
# Replace current values in the country column with the country values from the dictionary based on country_noc
world_indicators['Country Code'] = world_indicators['Country Code'].map(code_dict)
```

```python
# merge
country_data = pd.merge(
    country_medal_count,
    world_indicators,
    left_on=['country_noc', 'year'],
    right_on=['Country Code', 'Year'],
    how='left'
)
```

```python
# unique country codes in world_indicators
world_codes = world_indicators['Country Code'].unique()

# find missing country_noc values for each year in the merged dataset
missing_codes_all_years = country_data[~country_data['country_noc'].isin(world_codes)]['country_noc'].unique()

# Print the missing country codes
len(missing_codes_all_years)
```
